```python
import pandas as pd
import altair as alt
import ipywidgets as widgets
from IPython.display import display, clear_output

alt.data_transformers.enable('vegafusion')

# Load the file
cat_df = pd.read_csv('data/CAT-current-UBCBG.csv')

# Rename relevant columns
cat_df = cat_df.rename(columns={
    'TaxonName': 'Taxon',
    'ProvenanceCode': 'ProvenanceGroup',
    'ClimateRating_current': 'Current',
    'ClimateRating_emissions-limited_2050': '2050',
    'ClimateRating_business-as-usual_2090': '2090',
    'ClimateRating_bau-plus-1-degree_2090': '2090_plus_1deg'
})

# Fold Era columns into long format
long_df = cat_df.melt(
    id_vars=['ItemAccNoFull', 'ItemLocationCode', 'LocationCoordX', 'LocationCoordY', 
             'Taxon', 'LifeForm', 'ProvenanceGroup', 'LocationName'],
    value_vars=['Current', '2050', '2090', '2090_plus_1deg'],
    var_name='Era',
    value_name='ClimateRating'
)

# Filter to Main Garden only (exclude Nursery and Nitobe)
garden_df = long_df[
    (~long_df['LocationName'].str.contains('Nursery', na=False)) &
    (~long_df['LocationName'].str.contains('Nitobe', na=False))
].dropna(subset=['LocationCoordX', 'LocationCoordY'])

# Map LifeForm groups
def map_lifeform(lifeform):
    if pd.isna(lifeform):
        return 'Unknown'
    if lifeform in ['Tree', 'Shrub or Tree']:
        return 'Trees'
    if lifeform == 'Herbaceous Perennial':
        return 'Perennials'
    if lifeform == 'Annual':
        return 'Annuals'
    if lifeform == 'Bulb, Corm, or Tuber':
        return 'Bulbs'
    if lifeform in ['Shrub', 'Climber_Liana_Vine']:
        return 'Woody'
    if lifeform in ['Annual', 'Herbaceous Perennial', 'Bulb, Corm, or Tuber']:
        return 'Herbaceous'
    return 'Other'

garden_df['LifeFormGroup'] = garden_df['LifeForm'].apply(map_lifeform)
```

```python
import pandas as pd
import altair as alt

# 1. Load the comprehensive proofed dataset
df = pd.read_csv('data/FullJoin3_with_climate_ratings-proofed.csv')

# 2. Rename columns to match the specific names required by your Altair widget
rename_dict = {
    'climate_rating_current_sheet': 'Current',
    'climate_rating_emissions_limited_2050': '2050',
    'climate_rating_business_as_usual_2090': '2090',
    'climate_rating_bau_plus_1degree_2090': '2090_plus_1_degree',
    'GenusSpecies': 'Taxon',  # Mapping 'GenusSpecies' to 'Taxon' for the tooltip
}

# 3. Standardize Coordinates (Fixing the historical swaps)
df['LocationCoordY_fixed'] = df[['LocationCoordX', 'LocationCoordY']].min(axis=1) # Longitude
df['LocationCoordX_fixed'] = df[['LocationCoordX', 'LocationCoordY']].max(axis=1) # Latitude

# 4. Filter Locations (Removing Nursery and Nitobe)
# This replaces the need for manual outlier removal later
exclude_locs = ['Nursery', 'Nitobe Memorial Garden', 'Nitobe']
df_clean = df[~df['LocationName'].isin(exclude_locs)].copy()

# Remove spatial artifacts east of the garden core
df_clean = df_clean[df_clean['LocationCoordY_fixed'] < -123.245].copy()

# 5. Select and Rename Columns
df_ready = df_clean.rename(columns=rename_dict)
df_ready['LocationCoordX'] = df_clean['LocationCoordX_fixed']
df_ready['LocationCoordY'] = df_clean['LocationCoordY_fixed']

# Select only the columns needed for the visualization
cols_to_keep = [
    'ItemAccNoFull', 'ItemLocationCode', 
    'LocationCoordX', 'LocationCoordY', 
    'LifeForm', 'Taxon', 
    'Current', '2050', '2090', '2090_plus_1_degree'
]
final_merged_df = df_ready[cols_to_keep].copy()

# 6. Apply LifeForm Mapping (Matching your notebook's logic)
lifeform_mapping = {
    'Shrub': 'Woody',
    'Tree': 'Woody',
    'Shrub or Tree': 'Woody',
    'Climber_Liana_Vine': 'Woody',
    'Herbaceous Perennial': 'Perennial',
    'Bulb, Corm, or Tuber': 'Perennial',
    'Annual': 'Short-lived',
    'Biennial': 'Short-lived',
    'Habit unknown': 'Unknown'
}
final_merged_df['LifeForm'] = final_merged_df['LifeForm'].map(lifeform_mapping).fillna('Unknown')

# 7. Pivot to Long Format (Melt) for Altair
long_format_df = pd.melt(
    final_merged_df,
    id_vars=['ItemAccNoFull', 'ItemLocationCode', 'LocationCoordX', 'LocationCoordY', 'Taxon', 'LifeForm'],
    value_vars=['Current', '2050', '2090', '2090_plus_1_degree'],
    var_name='Era',
    value_name='ClimateRating'
)

# Preview to confirm it's ready for the widget
print("Data ready for visualization:")
print(long_format_df.head())
```

```python
# Setup dropdown options

era_order = ['Current', '2050', '2090', '2090_plus_1deg']
lifeform_options = ['(All)', 'Trees', 'Perennials', 'Annuals', 'Bulbs', 'Woody', 'Herbaceous', 'Other']

era_widget = widgets.Dropdown(
    options=era_order,
    value='Current',
    description='Era:'
)

lifeform_widget = widgets.Dropdown(
    options=lifeform_options,
    value='(All)',
    description='LifeForm:'
)

provenance_widget = widgets.Dropdown(
    options=['(All)'] + sorted(garden_df['ProvenanceGroup'].dropna().unique()),
    value='(All)',
    description='Provenance:'
)

output = widgets.Output()
```

```python
# Manually set final Garden plot boundaries
manual_long_min = -123.2525
manual_long_max = -123.2402
manual_lat_min = 49.248
manual_lat_max = 49.256

current_chart = None

def update_plot(*args):
    global current_chart
    filtered = garden_df[
        (garden_df['Era'] == era_widget.value)
    ]

    if lifeform_widget.value != '(All)':
        filtered = filtered[filtered['LifeFormGroup'] == lifeform_widget.value]

    if provenance_widget.value != '(All)':
        filtered = filtered[filtered['ProvenanceGroup'] == provenance_widget.value]

    filtered = filtered.dropna(subset=['ClimateRating'])

    chart = alt.Chart(filtered).mark_circle(size=60).encode(
        x=alt.X('LocationCoordY:Q', scale=alt.Scale(domain=[manual_long_min, manual_long_max]), axis=alt.Axis(title='Longitude')),
        y=alt.Y('LocationCoordX:Q', scale=alt.Scale(domain=[manual_lat_min, manual_lat_max]), axis=alt.Axis(title='Latitude')),
        color=alt.Color('ClimateRating:Q', scale=alt.Scale(domain=[5, 11], range=['red', 'yellow', 'green'])),
        tooltip=['Taxon:N', 'ClimateRating:Q', 'LocationCoordX:Q', 'LocationCoordY:Q']
    ).properties(
        width=600,
        height=600,
        title=f"Climate Rating Map ({era_widget.value})"
    )

    current_chart = chart

    with output:
        clear_output(wait=True)
        if not filtered.empty:
            display(chart)
        else:
            display(alt.Chart(pd.DataFrame({'x':[], 'y':[]})).mark_point())

# Hook widgets
for widget in [era_widget, lifeform_widget, provenance_widget]:
    widget.observe(update_plot, names='value')

# Layout
display(widgets.HBox([widgets.VBox([era_widget, lifeform_widget, provenance_widget]), output]))

# Initial Plot
update_plot()
```
